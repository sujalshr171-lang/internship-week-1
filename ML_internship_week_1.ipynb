```python
#7.Calculator
a=int(input("Enter first number:"))
b=int(input("Enter second number:"))
o=input("Enter operation (+,-,*,/):")
if o=="+":
  print(a+b)
elif o=="-":
  print(a-b)
elif o=="*":
  print(a*b)
elif o=="/":
  if b==0:
    print("Zero division error")
  else:
    print(a/b)
else:
  print("Invalid operation")
```

```python
#Working with CSV and JSON files
#1.CSV-Comma Separated Variables
import csv
with open("intern.csv","w",newline="")as file:
  writer=csv.writer(file)
  writer.writerow(["Name","Age","Role"])
  writer.writerow(["Sujal",18,"ML intern"])
  writer.writerow(["Simran",19,"ML intern"])
```

```python
with open("intern.csv","r")as file:
  reader=csv.reader(file)
  for row in reader:
      print(row)
```

```python
#2. JSON file
import json
intern={
    "name":"Sujal Shrestha",
    "age":18,
    "role":"ML intern"
}
with open("intern.json","w")as file:
  json.dump(intern,file,indent=4)
```

```python
with open("intern.json","r")as file:
  data=json.load(file)
print(data)
print(data["name"])
print(data["age"])
print(data["role"])
```
